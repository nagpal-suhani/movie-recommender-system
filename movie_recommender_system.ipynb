#### we are making a content based recommender system 
for which we need to create tags\
for each column if it does not contribute to tags we eliminate it\
relevant columns - genres, keywords, id, title, overview, release_date, cast, crew

```python
movies = movies[['genres', 'id','keywords', 'overview', 'title', 'cast', 'crew']]
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
# convert this [{"id": 28, "name": "Action"},{"id": 12, "name": "Adventure"},{"id": 14, "name": "Fantasy"},{"id": 878, "name": "Science Fiction"}]
# to ['Action', 'Adventure', 'Fantasy', Science Fiction]
```

```python
import ast
def convert(obj):
    l = []
    for i in ast.literal_eval(obj):   # genres is in string form rather than list so we first convert using ast.literal_eval
        l.append(i['name'])
    return l
```

```python
convert(movies.iloc[0].genres)
```

```python
# replacing genres column and keywords columns with their tag lists using convert function
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.iloc[1].cast
```

```python
# required top 3 actors from each movies
def convert_cast(obj):
    l = []
    counter=0
    for i in ast.literal_eval(obj):   # genres is in string form rather than list so we first convert using ast.literal_eval
        if counter!=3:
            l.append(i['name'])
            counter+=1
        else:
            break
    return l
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies['crew'][0]
```

```python
def fetch_director(obj):
    l = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            l.append(i['name'])
    return l   
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies.head()
```

```python
# we need to remove spaces from between names to avoid discrepancy
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_movies = movies[['id','title','tags']]
```

```python
new_movies.head()
```

```python
new_movies['tags'] = new_movies['tags'].apply(lambda x: " ".join(x))
new_movies['tags'] = new_movies['tags'].apply(lambda x: x.lower())
```

```python
new_movies.head()
```

```python
# stemming
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[] 
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(new_movies['tags']).toarray() 
```

```python
vectors
```

```python
# features = list(cv.get_feature_names_out())
# for word in features:
#     print(word)
cv.get_feature_names_out()
```

```python
# this may result in some similar words being in the list like - abilities and ability 
# to avoid this we apply stemming
# stemming changes ['loved', 'loving', 'love'] to ['love', 'love', 'love']
```

```python
new_movies['tags'] = new_movies['tags'].apply(stem)
```

```python
# we made vectors so the recommended movie would be the movies with least vector angles
# we use cosine similarity to find degrees betweeen each vector
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
sorted(list(enumerate(similarity[0])), reverse=True, key= lambda x: x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = new_movies[new_movies['title']==movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key= lambda x: x[1])[1:6]
    for i in movies_list:
        rec = new_movies.iloc[i[0]].title
        print(rec)
```

```python
recommend('Cars 2')
```
